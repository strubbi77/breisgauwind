```python
import pandas as pd
import numpy as np

DWD_Daten = pd.read_csv("Wetter/produkt_klima_tag_17810101_20231231_02290.txt",sep=';',na_values=['-999',' -999','  -999','   -999']).fillna(0)
DWD_Daten['Datum']=pd.to_datetime(DWD_Daten['MESS_DATUM'],format='%Y%m%d')
#DWD_Daten=DWD_Daten.set_index('Datum')
```

```python
df=DWD_Daten
```

```python
df['Jahr']=df['Datum'].dt.year
#df=df.drop(columns=['Date','Datum von','Datum bis','Jahr','Datum','Preis'], axis=1)
df_jahr=df.groupby(['Jahr']).mean(numeric_only=True)
```

```python
df_jahr
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()

val='  NM'

df_fi=df_jahr[df_jahr[val]!=0]

x=df_fi.index
y=df_fi[val]
#c=pm_day['color']
scatter=ax.scatter(x,y)
plt.show()
```

```python
df['Jahr']=df['Datum'].dt.year
df['Monat']=df['Datum'].dt.month

fig=plt.figure(figsize=(12,6))
fig.tight_layout()
st = fig.suptitle("Temperaturdurchschnitt über die Jahre")

for monat in range(1,13):
    ax=fig.add_subplot(3,4,monat)

    df_filter=df[df['Datum'].dt.month==monat]
    df_jahr=df_filter.groupby(['Jahr']).mean(numeric_only=True)
    x=df_jahr.index
    y=df_jahr[' TMK']
    #c=pm_day['color']
    scatter=ax.scatter(x,y)
    ax.annotate('{}'.format(monat),xy=(0,1), xycoords='axes fraction',xytext=(+0.5, -0.5), textcoords='offset fontsize',fontsize='medium', verticalalignment='top')
        
#df=df.drop(columns=['Date','Datum von','Datum bis','Jahr','Datum','Preis'], axis=1)
plt.show()
```
